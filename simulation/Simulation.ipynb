```python
import numpy as np

order = 20                  
chain_length = 100000
states = [0, 1]

context_probabilities = {}

def get_distribution_for_context(context):
    """
    Given a context (a tuple of length 'order'), return a probability distribution over the states.
    If this context is not already present in the dictionary, generate a random distribution and store it.
    """
    if context not in context_probabilities:
        probs = np.random.rand(len(states))
        probs /= np.sum(probs)
        context_probabilities[context] = probs
    return context_probabilities[context]

initial_context = tuple(np.random.choice(states, size=order))
chain = list(initial_context)

for _ in range(chain_length - order):
    current_context = tuple(chain[-order:])
    distribution = get_distribution_for_context(current_context)
    next_state = np.random.choice(states, p=distribution)
    chain.append(next_state)

output_filename = "20-order_markov_chain.txt"
with open(output_filename, "w") as f:
    f.write(" ".join(map(str, chain)))

print(f"Simulation complete. The chain of {chain_length} symbols has been saved to '{output_filename}'.")

```

```python
import numpy as np

order = 20                    
observed_length = 100_000     
total_hidden_length = observed_length + order  

X = np.zeros(total_hidden_length, dtype=np.int8
X[0:order] = np.random.randint(0, 2, size=order)

for i in range(order, total_hidden_length):
    X[i] = (X[i - 1] + X[i - order]) % 2

S = np.zeros(observed_length, dtype=np.int8)

noise = np.random.binomial(1, 0.1, size=observed_length)

for j in range(observed_length):
    S[j] = (X[j + order - 1] + X[j] + noise[j]) % 2

output_filename = "20-order_hidden_markov_chain.txt"
with open(output_filename, "w") as f:
    f.write(" ".join(map(str, S.tolist())))

print(f"Simulation complete. The observed Hidden Markov sequence of {observed_length} symbols has been saved to '{output_filename}'.")
```
